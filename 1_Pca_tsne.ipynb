```python
X = df[data.feature_names]

y = df['diagnosis']

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

#X = Independent variables (30 input features).
#y = Dependent variable (diagnosis).
```

```python
df['diagnosis_name'] = df['diagnosis'].map({
    0: 'Malignant',
    1: 'Benign'
})

df['diagnosis_name'].value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Scaled Training Shape:", X_train_scaled.shape)
print("Scaled Testing Shape:", X_test_scaled.shape)
```

```python
X_train_scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X.columns
)

X_train_scaled_df.head()
```

```python
print("Mean of scaled features:")
print(X_train_scaled_df.mean().round(2))

print("\nStandard deviation of scaled features:")
print(X_train_scaled_df.std().round(2))
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

print("Original shape:", X_train_scaled.shape)
print("PCA shape:", X_train_pca.shape)
```

```python
explained_variance = pca.explained_variance_ratio_

for i, variance in enumerate(explained_variance):
    print(f"PC{i+1}: {variance:.4f}")
```

```python
#A scree plot helps us understand how much variance each principal component explains.
plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(explained_variance) + 1),
    explained_variance,
    marker='o'
)

plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('PCA Scree Plot')
plt.grid()

plt.show()
```

```python
#It tells us how much total information is retained when we combine multiple principal components.
cumulative_variance = np.cumsum(explained_variance)

for i, variance in enumerate(cumulative_variance):
    print(f"First {i+1} Components: {variance:.4f}")

plt.figure(figsize=(10, 5))

plt.plot(
    range(1, 31),
    cumulative_variance,
    marker='o',
    color='green'
)

plt.axhline(
    y=0.95,
    color='red',
    linestyle='--',
    label='95% Variance'
)

plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance')
plt.legend()
plt.grid()

plt.show()
```

```python
#how many components are needed to preserve 95% of the variance.
n_components = np.argmax(cumulative_variance >= 0.95) + 1

print("Components required for 95% variance:", n_components)
```

```python
pca_95 = PCA(n_components=0.95)

X_train_pca_95 = pca_95.fit_transform(X_train_scaled)

X_test_pca_95 = pca_95.transform(X_test_scaled)

print("Original Features:", X_train_scaled.shape[1])

print("Reduced Features:", X_train_pca_95.shape[1])

print("Variance Retained:",
      round(sum(pca_95.explained_variance_ratio_), 4))
```

```python
pca_2 = PCA(n_components=2)

X_train_pca_2 = pca_2.fit_transform(X_train_scaled)

X_test_pca_2 = pca_2.transform(X_test_scaled)

print("Original Shape:", X_train_scaled.shape)

print("Reduced Shape:", X_train_pca_2.shape)
```

```python
pca_df = pd.DataFrame(
    X_train_pca_2,
    columns=['PC1', 'PC2']
)

pca_df['Diagnosis'] = y_train.map({
    0: 'Malignant',
    1: 'Benign'
}).values

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x='PC1',
    y='PC2',
    hue='Diagnosis',
    palette='Set1',
    s=60
)

plt.title('PCA Visualization - Breast Cancer Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')

plt.show()
#Separation between groups suggests that PCA has captured useful variation related to diagnosis
```

```python
print("========== PCA SUMMARY ==========")
print("Original Features:", X_train_scaled.shape[1])
print("Components for 95% variance:",
      pca_95.n_components_)
print("Variance retained by PCA 95%:",
      round(sum(pca_95.explained_variance_ratio_) * 100, 2), "%")
print("Components for visualization:", 2)
print("Variance retained by 2 components:",
      round(sum(pca_2.explained_variance_ratio_) * 100, 2), "%")
print("=================================")
```

```python
from sklearn.feature_selection import SelectKBest, f_classif
selector = SelectKBest(
    score_func=f_classif,
    k=10
)

X_train_kbest = selector.fit_transform(
    X_train_scaled,
    y_train
)

X_test_kbest = selector.transform(
    X_test_scaled
)

print("Original Features:", X_train_scaled.shape[1])
print("Selected Features:", X_train_kbest.shape[1])
```

```python
selected_mask = selector.get_support()

selected_features_kbest = X.columns[selected_mask]

print("Selected Features:")
print(selected_features_kbest.tolist())
```

```python
feature_scores = pd.DataFrame({
    'Feature': X.columns,
    'Score': selector.scores_
})

feature_scores = feature_scores.sort_values(
    by='Score',
    ascending=False
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=feature_scores.head(10),
    x='Score',
    y='Feature',
    hue='Feature',
    palette='viridis',
    legend=False
)

plt.title('Top 10 Features using SelectKBest')
plt.show()
#Features with higher ANOVA F-scores show stronger differences between the two diagnosis classes.
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
def evaluate_model(X_train_data, X_test_data,
                   y_train, y_test, method):

    model = LogisticRegression(
        max_iter=5000,
        random_state=42
    )

    model.fit(X_train_data, y_train)

    y_pred = model.predict(X_test_data)

    return {
        'Method': method,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1 Score': f1_score(y_test, y_pred)
    }
results = []

results.append(
    evaluate_model(
        X_train_scaled,
        X_test_scaled,
        y_train,
        y_test,
        'Original Features (30)'
    )
)
results.append(
    evaluate_model(
        X_train_pca_95,
        X_test_pca_95,
        y_train,
        y_test,
        'PCA (95% Variance)'
    )
)
results.append(
    evaluate_model(
        X_train_rfe,
        X_test_rfe,
        y_train,
        y_test,
        'RFE (10)'
    )
)
results_df = pd.DataFrame(results)

display(
    results_df.round(4)
)
metrics = ['Accuracy', 'Precision', 'Recall', 'F1 Score']

results_df.set_index('Method')[metrics].plot(
    kind='bar',
    figsize=(13, 6)
)

plt.title('Performance Comparison of Different Methods')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.xticks(rotation=30, ha='right')
plt.legend(loc='lower right')

plt.tight_layout()
plt.show()
best_method = results_df.loc[
    results_df['Accuracy'].idxmax()
]

print("Best Performing Method:")
print(best_method)
```
