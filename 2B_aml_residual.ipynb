```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python

# Number of data points
n = int(input("Enter number of data points: "))

# Take X values from user
X = list(map(float, input("Enter X values separated by space: ").split()))

# Take Y values from user
Y = list(map(float, input("Enter Y values separated by space: ").split()))
```

```python
# Convert to numpy arrays
X = np.array(X).reshape(-1, 1)
Y = np.array(Y)

# Check number of values
if len(X) != n or len(Y) != n:
    print("Error: Number of X and Y values must be equal to n.")
    exit()

# Linear Regression
model = LinearRegression()
model.fit(X, Y)
```

```python

# Slope and intercept
b = model.coef_[0]
a = model.intercept_

print("\nLinear Regression Equation:")
print(f"Y = {a:.3f} + {b:.3f}X")
```

```python

# Predicted values
Y_pred = model.predict(X)

# Residuals
residuals = Y - Y_pred
```

```python
# Display results
print("\nX\tActual Y\tPredicted Y\tResidual")

for x, y, yp, r in zip(X.flatten(), Y, Y_pred, residuals):
    print(f"{x:.2f}\t{y:.2f}\t\t{yp:.3f}\t\t{r:.3f}")

# Residual Plot
plt.scatter(X, residuals, color="blue", label="Residuals")

# Zero residual line
plt.axhline(y=0, color="red", linestyle="--", label="Zero Residual")

plt.xlabel("X")
plt.ylabel("Residual (Y - Ŷ)")
plt.title("Residual Plot")
plt.legend()
plt.grid(True)

plt.show()
```
